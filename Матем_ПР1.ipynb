1.2 Найдите: a) rref (Reduced Row Echelon Form) расширенной матрицы и б) общее решение системы линейных уравнений
$$
\begin{cases}
5x_1−x_2 + 5x_3 + 9x_4−2x_5 =−3, \\
3x_1 + 6x_3 + 9x_4−x_5 = 4, \\
2x_1 + 4x_3 + 6x_4−x_5 = 0.
\end{cases}
$$

```python
import numpy as np
import sympy as sp

# Расширенная матрица A|b
A = sp.Matrix([
    [5, -1, 5, 9, -2, -3],
    [3,  0, 6, 9, -1,  4],
    [2,  0, 4, 6, -1,  0]
])

# а) rref
rref_matrix, pivot_calls = A.rref()
print("а) rref(A|b) =")
sp.pprint(rref_matrix)
print()

# б) Общее решение системы с помощью символьных переменных
x1, x2, x3, x4, x5 = sp.symbols('x1 x2 x3 x4 x5', real=True)

# Уравнения (перенесли всё в левую часть: ... = 0)
eq1 = sp.Eq(5*x1 - x2 + 5*x3 + 9*x4 - 2*x5, -3)
eq2 = sp.Eq(3*x1 + 6*x3 + 9*x4 - x5, 4)
eq3 = sp.Eq(2*x1 + 4*x3 + 6*x4 - x5, 0)

# Решаем относительно свободных переменных
solution = sp.solve([eq1, eq2, eq3], [x1, x2, x5], dict=True)
print("б) Общее решение:")
sp.pprint(solution)
```

1.3. Пусть (x,y,z,u) – решение системы уравнений:
$$
\begin{cases}
32x+ 97y+ 90z+ 48u= 47 \\
35x+ 84y+ 91z+ 54u= 40 \\
17x+ 18y+ 21z+ 13u= 37
\end{cases}
$$
Объясните, почему yможно представить как функцию от xвида y= a+ bx, где a,b– константы, зависящие только от коэффициентов исходной системы уравнений. 
Найдите: 1) константу a; 
2)константу b.

**Ответ**: Система из 3 уравнений и 4 неизвестных совместна, значит одна переменная свободная. Выражая (y, z, u) через свободную x, получаем, что y — линейная функция от x, т.е. y = a + bx.

```python
import sympy as sp

x, y, z, u = sp.symbols('x y z u')

# Матрица коэффициентов и вектор правых частей
A = sp.Matrix([
    [32, 97, 90, 48],
    [35, 84, 91, 54],
    [17, 18, 21, 13]
])
b = sp.Matrix([47, 40, 37])

# 1. Обоснование: система из 3 уравнений с 4 неизвестными совместна,
#    значит, у неё есть свободная переменная. Берём x как свободную
#    и выражаем остальные через x.
B = A[:, 1:]               # столбцы при (y, z, u)
rhs = b - A[:, 0] * x      # правая часть минус вклад x
sol = B.LUsolve(rhs)

# 2. y как функция от x
y_expr = sp.simplify(sol[0])
print("y =", y_expr)

# 3. Константы
a = sp.simplify(sp.Poly(sp.expand(y_expr), x).coeff_monomial(sp.S(1)))
b_const = sp.simplify(sp.Poly(sp.expand(y_expr), x).coeff_monomial(x))
print("a =", a)
print("b =", b_const)
```

a) a_ij = min(i,j); б) a_ij = max(i,j); в)∗ a_ij= |i−j|.

```python
import sympy as sp

def det_min(n):
    return sp.Matrix([[min(i, j) for j in range(1, n+1)] for i in range(1, n+1)]).det()

def det_max(n):
    return sp.Matrix([[max(i, j) for j in range(1, n+1)] for i in range(1, n+1)]).det()

def det_abs(n):
    return sp.Matrix([[abs(i - j) for j in range(1, n+1)] for i in range(1, n+1)]).det()

# Проверка для нескольких значений n
for n in range(1, 6):
    print(f"n={n}:  min={det_min(n)},  max={det_max(n)},  |i-j|={det_abs(n)}")
```

1.6 Найдите наибольшее значение определителя шестого порядка, составленного из чисел−1, 0, и 1. (Можно перебором).

```python
import numpy as np

def improve(M):
    """Локальное улучшение: перебираем каждый элемент, пробуем -1/0/1,
    оставляем, если определитель вырос."""
    n = M.shape[0]
    best = abs(int(round(np.linalg.det(M))))
    for _ in range(50):          # несколько проходов
        for i in range(n):
            for j in range(n):
                orig = M[i, j]
                for v in [-1, 0, 1]:
                    if v == orig:
                        continue
                    M[i, j] = v
                    d = abs(int(round(np.linalg.det(M))))
                    if d > best:
                        best = d
                    else:
                        M[i, j] = orig      # откатываем
    return best, M

best = 0
best_mat = None

for trial in range(5000):        # много случайных стартов
    M = np.random.choice([-1, 0, 1], size=(6, 6)).astype(float)
    d, M2 = improve(M)
    if d > best:
        best = d
        best_mat = M2.copy()
        print("Новый рекорд:", best)     # покажет прогресс
    if best == 160:              # известный максимум для n=6
        break

print("\nИтоговый максимум =", best)
print("Матрица (как числа):")
print(best_mat.astype(int))
print("Проверка det =", int(round(np.linalg.det(best_mat))))
```

1.7 Матрица A= (ai,j ) имеет размер n×n, где n= 80000. Известно, что ее диагональные элементы образуют арифметическую прогрессию с начальным элементом a1,1 = 0,99 и конечным элементом an,n = 1,01. Также известно, что все элементы матрицы A, расположенные вне главной диагонали, равны нулю за исключением двух элементов: a1,2 = 0,57 и a2,1 = 0,03. 

Найдите: 1) определитель матрицы A; 2) след матрицы, обратной к A.


```python
import numpy as np

n = 80000

# 1) Диагональные элементы: арифметическая прогрессия от 0.99 до 1.01
d = np.linspace(0.99, 1.01, n)   # n точек, от 0.99 (i=1) до 1.01 (i=n)

# --- Определитель ---
# Матрица почти диагональная. Единственное возмущение — блок 2x2 в левом верхнем углу:
#   [[d0, 0.57],
#    [0.03, d1]]
# Значит  det(A) = (d0*d1 - 0.57*0.03) * prod(d[2:])

det_2x2 = d[0]*d[1] - 0.57*0.03
det_rest = np.prod(d[2:])         # произведение остальных диагональных элементов
det_A = det_2x2 * det_rest
print("1) det(A) =", round(det_A, 6))

# --- След обратной матрицы ---
# Диагональные элементы A^{-1}:
#   * позиции 1,2 — из инверсии блока 2x2
#   * остальные — 1/d[i]
# trace(A^{-1}) = (a+d)/(ad-bc)  [из блока]  +  sum(1/d[i], i=3..n)

a, b, c, dd = d[0], 0.57, 0.03, d[1]
det_blk = a*dd - b*c
trace_blk = (a + dd) / det_blk          # диагональные элементы инверсии 2x2 блока
trace_rest = np.sum(1.0 / d[2:])        # остальные 1/d_i
trace = trace_blk + trace_rest
print("2) trace(A^{-1}) =", round(trace, 3))
```

1.8 Матрица A= (ai,j ) имеет размер n×n, где n= 43000. 
Известно, что: 
1) диагональные элементы образуют геометрическую прогрессию с начальным элементом a1,1 = 1 и конечным элементом an,n = 1,07; 
2) элементы первой строки образуют арифметическую прогрессию с конечным элементом a1,n = 1,10; 
3) все элементы матрицы A, расположенные вне главной диагонали и вне первой строки, равны нулю. 

Найдите обратную матрицу B= A−1 и укажите в ответе: 
1) наименьший элемент матрицы B; 
2) сумму элементов матрицы B.


```python
import numpy as np

n = 43000
m = n - 1

# Диагональ: геометрическая прогрессия 1 -> 1.07  (d_i = q^(i-1), q^(n-1)=1.07)
q = 1.07 ** (1.0 / m)
i = np.arange(1, n + 1)
d = q ** (i - 1)

# Первая строка: арифметическая прогрессия 1 -> 1.10
r = 1 + (i - 1) * (0.10 / m)

# Матрица B = A^-1 имеет вид:
#   B[0,0] = 1,  B[0,j] = -r_j/d_j (j>=2),  B[i,i] = 1/d_i (i>=2)
offdiag  = -r[1:] / d[1:]
diag_inv =  1.0  / d[1:]

# 1) наименьший элемент
print("1) наименьший элемент =", round(np.min(np.concatenate([[1.0], offdiag, diag_inv])), 3))

# 2) сумма элементов
print("2) сумма элементов =", round(1.0 + np.sum(diag_inv) - np.sum(r[1:] / d[1:]), 1))
```

1.9 Даны матрицы:
$$
I = \begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix},
\qquad
A = \begin{pmatrix} 7\gamma & 14\gamma \\ 9\gamma & 1\gamma \end{pmatrix},
\qquad \gamma = 0{,}057.
$$
Пусть 
$$
S_{m,n} = \sum_{i=m}^{n} A^{i}, \qquad
B_m = (I - A)^{-1} A^{m}.
$$
1) Найдите наибольший элемент матрицы S= S4,70. 
2) Найдите наибольший элемент матрицы B= B4. 
3) Найдите наименьшее k, такое, что при любом n > kнаибольший элемент матрицы Sk,n меньше 0,0002.

```python
import numpy as np

gamma = 0.057
A = np.array([[7*gamma, 14*gamma],
              [9*gamma, 1*gamma]])
I = np.eye(2)

def m(M):
    return np.max(M)

# 1) S_{4,70} = sum_{i=4}^{70} A^i
S = sum(np.linalg.matrix_power(A, i) for i in range(4, 71))
print("1) max S4,70 =", round(m(S), 6))

# 2) B4 = (I-A)^{-1} * A^4
A4 = np.linalg.matrix_power(A, 4)
B4 = np.linalg.inv(I - A) @ A4
print("2) max B4 =", round(m(B4), 6))

# 3) B_k = (I-A)^{-1} A^k  -> предел S_{k,n} при n->беск.
#    нужен наименьший k c max B_k < 0.0002 (тогда верно для всех n)
B_inv = np.linalg.inv(I - A)
k = 0
while m(B_inv @ np.linalg.matrix_power(A, k)) >= 0.0002:
    k += 1
print("3) наименьшее k =", k)
```

1.14 Найдите минимальный многочлен матрицы
$$
A=
\begin{pmatrix}
−3 & 7 & −6 & −1 & 8 & −3 & −2 \\
1 & −5 & 8 & −3 & −5 & 4 & −5 \\
1 & 3 & −4 & 6 & −2 & 0 & 8 \\
−1 & 11 & −12 & 9 & 3 & −3 & 9 \\
−3 & 11 & −12 & 5 & 8 & −4 & 6 \\
−3 & 2 & 1 & −4 & 4 & 1 & −7 \\
1 & −5 & 6 & −2 & −4 & 3 & −2
\end{pmatrix}
$$

```python
import numpy as np
import sympy as sp

A = np.array([
    [-3,  7, -6, -1,  8, -3, -2],
    [ 1, -5,  8, -3, -5,  4, -5],
    [ 1,  3, -4,  6, -2,  0,  8],
    [-1, 11,-12,  9,  3, -3,  9],
    [-3, 11,-12,  5,  8, -4,  6],
    [-3,  2,  1, -4,  4,  1, -7],
    [ 1, -5,  6, -2, -4,  3, -2]
], dtype=float)

n = A.shape[0]

# ---- Метод Крылова: локальный минимальный многочлен по вектору v ----
def krylov(v):
    for m in range(1, n + 1):
        M = np.column_stack([np.linalg.matrix_power(A, i) @ v for i in range(m + 1)])
        if np.linalg.matrix_rank(M) <= m:          # нашли линейную зависимость
            Mb, col = M[:, :-1], M[:, -1]
            coef, *_ = np.linalg.lstsq(Mb, col, rcond=None)   # A^m v = sum coef_i A^i v
            coeffs = [sp.Rational(str(round(c, 8))).limit_denominator() for c in coef] + [sp.Integer(1)]
            return sp.factor(sum(coeffs[i] * x**i for i in range(m + 1)))

# НОК локальных многочленов по нескольким векторам = минимальный многочлен
minp = sp.Integer(1)
for seed in range(3):
    v = np.random.default_rng(seed).standard_normal(n)
    minp = sp.lcm(minp, krylov(v))

print("\nМинимальный многочлен матрицы A:")
print(sp.factor(minp))
```

```python
import numpy as np

# Матрица самосопряжённого оператора
A = np.array([
    [17, -8,  4],
    [-8, 17, -4],
    [ 4, -4, 11]
], dtype=float)

# 1) Собственные значения и векторы
vals, vecs = np.linalg.eigh(A)   # eigh — для симметричных матриц, даёт ортонормированный базис

print("Собственные значения: λ =", np.round(vals, 6))
print("\nСобственные векторы (столбцы, уже ортонормированные):")
print(np.round(vecs, 6))

# Проверка ортонормированности: V^T V = I
print("\nПроверка ортонормированности V^T·V =")
print(np.round(vecs.T @ vecs, 6))

# 2) Матрица оператора в собственном базисе (диагональная)
D = vecs.T @ A @ vecs
print("\nМатрица оператора в собственном базисе (V^T·A·V):")
print(np.round(D, 6))
```

1.17 Определите тип кривой второго порядка, её каноническое уравнение и найдите каноническую систему координат, если кривая задана уравнением:
$$
a)  5x^{2} + 4xy+ 8y^{2}−32x−56y+ 80 = 0; \\
b)  5x^{2} + 12xy−22x−12y−19 = 0.
$$

```python
import sympy as sp

def conic(a11, a12, a22, a13, a23, a33, label):
    x, y = sp.symbols('x y')
    A = sp.Matrix([[a11, a12], [a12, a22]])

    print(f"--- {label} ---")
    print("Квадратичная матрица A:\n", A)

    # собственные значения
    lam = A.eigenvals()
    keys = sorted(lam.keys())
    print("Собственные значения:", [(sp.nsimplify(k), lam[k]) for k in keys])

    s = [float(k) for k in keys]
    t = ("ЭЛЛИПС" if all(v > 0 for v in s)
         else "ЭЛЛИПС (мнимый)" if all(v < 0 for v in s)
         else "ПАРАБОЛА" if any(v == 0 for v in s)
         else "ГИПЕРБОЛА")
    print("Тип кривой:", t)

    # F(x,y) и центр: решаем grad(F) = 0
    F = a11*x**2 + 2*a12*x*y + a22*y**2 + 2*a13*x + 2*a23*y + a33
    sol = sp.solve([sp.Eq(sp.diff(F, x), 0), sp.Eq(sp.diff(F, y), 0)], [x, y], dict=True)
    if sol:
        cx, cy = sol[0][x], sol[0][y]
        Fc = sp.simplify(F.subs({x: cx, y: cy}))
        print("Центр:", (cx, cy))
        print("F(центр) =", Fc)
    else:
        cx, cy = None, None
        print("Центра нет (парабола или вырожденный случай)")

    # главные направления
    evec = A.eigenvects()
    vecs = {k: vlist[0] for k, mult, vlist in evec}
    print("Главные направления:")
    for k in keys:
        print(f"  для λ={sp.nsimplify(k)}: {sp.simplify(vecs[k])}")
    print("-" * 40)

conic(5, 2, 8, -16, -28, 80, "а) 5x^2+4xy+8y^2-32x-56y+80=0");
conic(5, 6, 0, -11, -6, -19, "б) 5x^2+12xy-22x-12y-19=0");
```

1.18 При каких значениях параметра λ квадратичная форма Q(x) является положительно определенной, если
$$
Q(x_1,x_2,x_3) =x^{2}_1 + 4x^{2}_2 + x^{2}_3 + 2λx_1x_2 + 10x_1x_3 + 6x_2x_3.
$$

```python
import sympy as sp

lam = sp.symbols('lambda')

A = sp.Matrix([
    [1,  lam, 5],
    [lam, 4,  3],
    [5,   3,  1]
])

# Угловые миноры
D2 = sp.Matrix([[1, lam], [lam, 4]]).det()
D3 = A.det()

print("Δ2 =", sp.factor(D2))
print("Δ3 =", sp.factor(D3))

# Решаем систему неравенств напрямую (без ручного пересечения множеств)
sol = sp.reduce_inequalities(sp.And(D2 > 0, D3 > 0), [lam])
print("\nУсловия Сильвестра Δ2>0 и Δ3>0 одновременно:")
print(sol)
```

1.19 При каких значениях параметра λ квадратичная форма Q(x) является отрицательно определенной, если
$$
Q(x_1,x_2,x_3) =λx^{2}_1−2x^{2}_2−3x^{2}_3 + 2x_1x_2−2x_1x_3 + 2x_2x_3.
$$

```python
import sympy as sp

lam = sp.symbols('lambda')

D1 = lam
D2 = -2*lam - 1
D3 = 5*lam + 3

i1 = sp.solve_univariate_inequality(D1 < 0, lam)
i2 = sp.solve_univariate_inequality(D2 > 0, lam)
i3 = sp.solve_univariate_inequality(D3 < 0, lam)

print("Δ1 < 0:", i1)
print("Δ2 > 0:", i2)
print("Δ3 < 0:", i3)

res = i1 & i2 & i3
print("\nПересечение всех условий:", res)
```

1.20 Эквивалентны ли над полем вещественных чисел квадратичные формы Q1 и Q2, где
$$
Q_1= 2x^{2}_1 + 9x^{2}_2 + 3x^{2}_3 + 8x_1x_2−4x_1x_3−10x_2x_3; \\
Q_2= 2x_1 + 3x^{2}_2 + 6x^{2}_3−4x_1x_2−4x_1x_3 + 8x_2x_3
$$

```python
import sympy as sp

# Матрицы квадратичных форм
A1 = sp.Matrix([
    [ 2,  4, -2],
    [ 4,  9, -5],
    [-2, -5,  3]
])

A2 = sp.Matrix([
    [ 2, -2, -2],
    [-2,  3,  4],
    [-2,  4,  6]
])

def signature(A):
    """Вычисляет сигнатуру (p, q, r) квадратичной формы по собственным значениям."""
    p = q = r = 0
    for val in A.eigenvals().keys():
        v = sp.simplify(val)
        if v > 0:
            p += 1
        elif v < 0:
            q += 1
        else:
            r += 1
    return p, q, r

# Вычисляем сигнатуры
sig1 = signature(A1)
sig2 = signature(A2)

print("Сигнатура Q1 (p, q, r) =", sig1)
print("Сигнатура Q2 (p, q, r) =", sig2)

# Эквивалентность над R: совпадение сигнатур (закон инерции Сильвестра)
equivalent = (sig1 == sig2)
print("\nСигнатуры совпадают:", equivalent)
if equivalent:
    print("=> Q1 и Q2 ЭКВИВАЛЕНТНЫ над R")
else:
    print("=> Q1 и Q2 НЕ эквивалентны над R")
```
